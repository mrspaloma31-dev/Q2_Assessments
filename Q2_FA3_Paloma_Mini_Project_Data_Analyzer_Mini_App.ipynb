```python
def display_scores(scores, title="Current Student Scores"):
    """Traverses and displays the list elements with their indices."""
    print(f"\n--- {title} ---")
    if not scores:
        print("No scores available.")
        return

    for index, score in enumerate(scores):
        print(f"Student [{index}]: {score}")
    print("-" * (len(title) + 8))


def search_score(scores):
    """Searches for a specific score in the list and reports its index."""
    try:
        target = float(input("\nEnter the score to search for: "))
        found_indices = [i for i, score in enumerate(scores) if score == target]

        if found_indices:
            print(f"Score {target} found at index(es): {found_indices}")
        else:
            print(f"Score {target} was not found in the list.")
    except ValueError:
        print("Invalid input. Please enter a numeric score.")


def compute_average(scores):
    """Computes and displays the average, highest, and lowest scores."""
    if not scores:
        print("\nCannot compute statistics: The score list is empty.")
        return

    total = sum(scores)
    average = total / len(scores)
    print("\n--- Score Computations ---")
    print(f"Total Students : {len(scores)}")
    print(f"Average Score  : {average:.2f}")
    print(f"Highest Score  : {max(scores)}")
    print(f"Lowest Score   : {min(scores)}")


def update_score(scores):
    """Updates a specific value in the list using its index."""
    display_scores(scores, "Select a Student Index to Update")
    try:
        index = int(input("Enter the student index to update: "))
        if 0 <= index < len(scores):
            new_score = float(input(f"Enter the new score for Student [{index}] (Current: {scores[index]}): "))
            old_score = scores[index]
            scores[index] = new_score
            print(f"Success: Updated Student [{index}] score from {old_score} to {new_score}.")
            return True
        else:
            print(f"Error: Index out of range. Please choose between 0 and {len(scores) - 1}.")
    except ValueError:
        print("Invalid input. Please enter valid numbers.")
    return False


def main():
    """Main program loop providing the text-based user interface."""
    # Initial dataset stored in a list
    student_scores = [85.0, 92.5, 78.0, 88.5, 95.0, 81.0]
    has_updates = False

    while True:
        print("\n=== Student Score Analyzer ===")
        print("Select the number of the operation you want to perform:")
        print("1. Display Scores")
        print("2. Search Score")
        print("3. Compute Average Score")
        print("4. Update Score")
        print("5. Display Updated Scores")
        print("6. Exit")

        choice = input("Enter your choice: ").strip()

        if choice == '1':
            display_scores(student_scores, "Student Scores")
        elif choice == '2':
            search_score(student_scores)
        elif choice == '3':
            compute_average(student_scores)
        elif choice == '4':
            if update_score(student_scores):
                has_updates = True
        elif choice == '5':
            if has_updates:
                display_scores(student_scores, "Updated Student Scores")
            else:
                print("\nNo modifications have been made yet.")
                display_scores(student_scores, "Current Student Scores")
        elif choice == '6':
            print("\nExiting Student Score Analyzer. Goodbye!")
            break
        else:
            print("\nInvalid choice. Please enter a number between 1 and 6.")


if __name__ == "__main__":
    main()
```
