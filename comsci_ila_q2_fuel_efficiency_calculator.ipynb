```python
def calculate_efficiency(distance, fuel):
    # Calculate kilometers per liter
    efficiency = distance / fuel
    return efficiency

# Ask user for inputs
distance_traveled = float(input("Enter distance traveled (in kilometers): "))
fuel_consumed = float(input("Enter fuel consumed (in liters): "))

# Run function and display output
result = calculate_efficiency(distance_traveled, fuel_consumed)
print(f"Fuel Efficiency: {result:.1f} km/L")
```
